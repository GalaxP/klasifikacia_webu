```python
def load_jsonl(filepath: str) -> pd.DataFrame:
    records: List[Dict[str, Any]] = []
    with open(filepath, "r", encoding="utf-8") as file:
        for line_number, line in enumerate(file, start=1):
            line = line.strip()
            if not line:
                continue
            try:
                records.append(json.loads(line))
            except json.JSONDecodeError:
                print(f"Preskakujem chybný JSON na riadku {line_number}.")
    return pd.DataFrame(records)

def parse_labels(value: Any) -> List[str]:
    if value is None:
        return []
    if isinstance(value, list):
        labels = value
    elif isinstance(value, str):
        value = value.strip()
        if not value:
            return []
        if value.startswith("["):
            try:
                parsed = json.loads(value)
                labels = parsed if isinstance(parsed, list) else [value]
            except json.JSONDecodeError:
                labels = re.split(r"[,;|]", value)
        else:
            labels = re.split(r"[,;|]", value)
    else:
        labels = [str(value)]

    cleaned_labels = []
    for label in labels:
        label = str(label).strip().lower()
        if label:
            cleaned_labels.append(label)
    return sorted(set(cleaned_labels))

def filter_rare_labels(df: pd.DataFrame, label_col: str, min_count: int) -> pd.DataFrame:
    label_counts = Counter(label for labels in df[label_col] for label in labels)
    valid_labels = {label for label, count in label_counts.items() if count >= min_count}

    df = df.copy()
    df[label_col] = df[label_col].apply(lambda labels: sorted(label for label in labels if label in valid_labels))
    df = df[df[label_col].map(len) > 0].reset_index(drop=True)

    print(f"Počet labelov po filtrovaní: {len(valid_labels)}")
    print(f"Odstránené labely s výskytom menším ako {min_count}.")
    return df

def multilabel_train_test_split(X: Sequence[Any], y: np.ndarray, test_size: float, random_state: int) -> Tuple[np.ndarray, np.ndarray]:
    indices = np.arange(len(X))
    try:
        from iterstrat.ml_stratifiers import MultilabelStratifiedShuffleSplit
        splitter = MultilabelStratifiedShuffleSplit(n_splits=1, test_size=test_size, random_state=random_state)
        train_idx, test_idx = next(splitter.split(np.zeros(len(indices)), y))
        return train_idx, test_idx
    except ImportError:
        print("Balík iterative-stratification nie je nainštalovaný. Používam obyčajný train_test_split.")
        train_idx, test_idx = train_test_split(indices, test_size=test_size, random_state=random_state)
        return train_idx, test_idx

def get_positive_class_proba(classifier: OneVsRestClassifier, X: Any) -> np.ndarray:
    probabilities = classifier.predict_proba(X)
    if isinstance(probabilities, list):
        probabilities = np.vstack([item[:, 1] for item in probabilities]).T
    return np.asarray(probabilities)

def tune_thresholds(y_true: np.ndarray, y_proba: np.ndarray, threshold_grid: np.ndarray) -> np.ndarray:
    thresholds = np.full(y_true.shape[1], 0.5, dtype=float)
    for label_idx in range(y_true.shape[1]):
        if y_true[:, label_idx].sum() == 0:
            continue
        best_threshold = 0.5
        best_score = -1.0
        for threshold in threshold_grid:
            y_pred_label = (y_proba[:, label_idx] >= threshold).astype(int)
            score = f1_score(y_true[:, label_idx], y_pred_label, zero_division=0)
            if score > best_score:
                best_score = score
                best_threshold = threshold
        thresholds[label_idx] = best_threshold
    return thresholds

def apply_thresholds(y_proba: np.ndarray, thresholds: np.ndarray, ensure_at_least_one_label: bool = True) -> np.ndarray:
    y_pred = (y_proba >= thresholds).astype(int)
    if ensure_at_least_one_label:
        empty_rows = np.where(y_pred.sum(axis=1) == 0)[0]
        if len(empty_rows) > 0:
            best_label_indices = np.argmax(y_proba[empty_rows], axis=1)
            y_pred[empty_rows, best_label_indices] = 1
    return y_pred

_LANGDETECT_AVAILABLE = None


def is_probably_czech(text: str) -> bool:
    global _LANGDETECT_AVAILABLE

    if not FILTER_NON_CZECH:
        return True

    if _LANGDETECT_AVAILABLE is False:
        return True

    try:
        from langdetect import DetectorFactory, detect_langs

        _LANGDETECT_AVAILABLE = True
        DetectorFactory.seed = RANDOM_STATE
        sample = text[:5000]
        detected_languages = detect_langs(sample)
        return any(
            item.lang == "cs" and item.prob >= MIN_CZECH_PROB
            for item in detected_languages
        )
    except ImportError:
        if _LANGDETECT_AVAILABLE is None:
            print(
                "Balík langdetect nie je nainštalovaný. "
                "Jazykový filter sa vypína."
            )
        _LANGDETECT_AVAILABLE = False
        return True
    except Exception:
        return False
```

```python
class TextProcessor:
    def __init__(self) -> None:
        self.stop_words = CZECH_STOPWORDS
        self.nlp = None

        if USE_STANZA_LEMMATIZATION:
            try:
                import os
                import stanza

                model_dir = os.path.join(os.path.expanduser("~"), ".stanza")
                cs_model_path = os.path.join(model_dir, "models", "cs")

                if not os.path.exists(cs_model_path):
                    print("Sťahujem český model pre Stanza...")
                    stanza.download("cs", verbose=False)

                self.nlp = stanza.Pipeline(
                    lang="cs",
                    processors="tokenize,pos,lemma",
                    use_gpu=False,
                    verbose=False,
                )
                print("Používam Stanza český model na lematizáciu.")
            except Exception as exc:
                print(
                    "Stanza český model sa nepodarilo načítať. "
                    f"Pokračujem bez lematizácie. Detail: {exc}"
                )
                self.nlp = None

    def decompress_html(self, value: Any) -> str:
        if value is None:
            return ""
        if not isinstance(value, str):
            value = str(value)
        value = value.strip()
        if not value:
            return ""
        if (
            "<html" in value.lower()
            or "<body" in value.lower()
            or "<div" in value.lower()
        ):
            return value

        try:
            decoded = base64.b64decode(value)
        except Exception:
            return value

        try:
            decompressed = gzip.decompress(decoded)
            return decompressed.decode("utf-8", errors="ignore")
        except Exception:
            try:
                return decoded.decode("utf-8", errors="ignore")
            except Exception:
                return ""

    def _extract_meta_content(
        self, soup: BeautifulSoup, names: Iterable[str]
    ) -> List[str]:
        values: List[str] = []
        for name in names:
            meta = soup.find("meta", attrs={"name": name})
            if meta and meta.get("content"):
                values.append(str(meta["content"]))
        return values

    def _drop_boilerplate_nodes(self, soup: BeautifulSoup) -> None:
        for comment in soup.find_all(
            string=lambda item: isinstance(item, Comment)
        ):
            comment.extract()

        for tag in soup(
            [
                "script",
                "style",
                "noscript",
                "svg",
                "canvas",
                "iframe",
                "form",
                "nav",
                "footer",
                "aside",
                "template",
                "dialog",
            ]
        ):
            tag.decompose()

        noisy_keywords = {
            "cookie",
            "cookies",
            "consent",
            "gdpr",
            "privacy",
            "newsletter",
            "subscribe",
            "popup",
            "modal",
            "banner",
            "advert",
            "ads",
            "promo",
            "breadcrumb",
            "login",
            "signin",
            "register",
            "cart",
            "basket",
            "menu",
            "sidebar",
            "footer",
            "header",
            "toolbar",
            "pagination",
            "sharing",
            "social",
            "filter",
            "sort",
            "search",
            "recommend",
            "related",
        }

        tags = list(soup.find_all(True))

        for tag in reversed(tags):
            if tag is None:
                continue

            attrs = getattr(tag, "attrs", None)
            if attrs is None:
                continue

            class_attr = tag.get("class", [])
            if isinstance(class_attr, (list, tuple)):
                class_text = " ".join(map(str, class_attr))
            else:
                class_text = str(class_attr or "")

            attrs_to_check = [
                str(tag.get("id", "") or ""),
                class_text,
                str(tag.get("role", "") or ""),
                str(tag.get("aria-label", "") or ""),
            ]
            attrs_text = " ".join(x.lower() for x in attrs_to_check if x)

            if any(keyword in attrs_text for keyword in noisy_keywords):
                tag.decompose()
                continue

            style = str(tag.get("style", "") or "").lower()
            if (
                "display:none" in style
                or "visibility:hidden" in style
                or "opacity:0" in style
            ):
                tag.decompose()

    def _extract_main_text_trafilatura(self, html_text: str) -> str:
        try:
            import trafilatura

            extracted = trafilatura.extract(
                html_text,
                include_comments=False,
                include_tables=False,
                include_formatting=False,
                include_links=False,
                favor_precision=True,
                deduplicate=True,
            )
            return extracted or ""
        except Exception:
            return ""

    def _deduplicate_lines(self, text: str) -> str:
        parts = [
            part.strip()
            for part in re.split(r"[\n\r]+|(?<=[.!?])\s{2,}", text)
            if part.strip()
        ]

        if not parts:
            return text

        counts = Counter(parts)
        cleaned_parts: List[str] = []

        for part in parts:
            if counts[part] > 3:
                continue
            cleaned_parts.append(part)

        return " ".join(cleaned_parts)

    def _remove_boilerplate_phrases(self, text: str) -> str:
        patterns = [
            r"používáme cookies[^.]{0,300}",
            r"tato stránka používá cookies[^.]{0,300}",
            r"používáme soubory cookies[^.]{0,300}",
            r"zásady ochrany osobních údajů[^.]{0,300}",
            r"ochrana osobních údajů[^.]{0,300}",
            r"nastavení soukromí[^.]{0,300}",
            r"personalizovan[ýaé] reklam[ay][^.]{0,300}",
            r"oprávněn[ýe] zájem[^.]{0,300}",
            r"využíváme vaše údaje[^.]{0,300}",
            r"spravovat souhlas[^.]{0,300}",
            r"kliknutím na[^.]{0,200}souhlas",
            r"přihl[aá]sit se",
            r"registrovat se",
            r"vložit do košíku",
            r"přidat do košíku",
            r"dostupnost na prodejně",
            r"doprava zdarma",
            r"seřadit podle",
            r"zobrazit více",
            r"další produkty",
            r"související produkty",
        ]

        cleaned = text
        for pattern in patterns:
            cleaned = re.sub(pattern, " ", cleaned, flags=re.IGNORECASE)

        return cleaned

    def clean_html(self, html_text: str) -> str:
        if not html_text:
            return ""

        soup = BeautifulSoup(html_text, "lxml")
        self._drop_boilerplate_nodes(soup)

        important_parts: List[str] = []

        if soup.title and soup.title.string:
            important_parts.extend([soup.title.string] * 3)

        meta_parts = self._extract_meta_content(
            soup,
            ["description", "keywords", "og:description"],
        )
        for part in meta_parts:
            important_parts.extend([part] * 2)

        for heading in soup.find_all(["h1", "h2", "h3"]):
            heading_text = heading.get_text(" ", strip=True)
            if heading_text:
                important_parts.extend([heading_text] * 2)

        main_text = self._extract_main_text_trafilatura(str(soup))
        if not main_text:
            main_candidates = []

            for selector in [
                "main",
                "article",
                "[role='main']",
                ".content",
                ".main",
                ".article",
                ".post",
                ".entry",
            ]:
                try:
                    selected = soup.select(selector)
                except Exception:
                    selected = []
                for node in selected:
                    txt = node.get_text(" ", strip=True)
                    if len(txt) >= 200:
                        main_candidates.append(txt)

            if main_candidates:
                main_text = max(main_candidates, key=len)
            else:
                main_text = soup.get_text(separator=" ", strip=True)

        text = " ".join(important_parts + [main_text])
        text = self._deduplicate_lines(text)
        text = self._remove_boilerplate_phrases(text)
        text = text.lower()

        text = re.sub(r"https?://\S+|www\.\S+", " ", text)
        text = re.sub(r"\S+@\S+", " ", text)
        text = re.sub(r"&nbsp;|&#160;", " ", text)
        text = re.sub(r"\b(cookie|cookies|gdpr|consent)\b", " ", text)
        text = re.sub(r"\b\d{1,2}:\d{2}\b", " ", text)
        text = re.sub(r"[^0-9a-záčďéěíňóřšťúůýž\s-]", " ", text)
        text = re.sub(r"\b[a-záčďéěíňóřšťúůýž]{1,2}\b", " ", text)
        text = re.sub(r"\s+", " ", text).strip()

        return text

    def preprocess_text(self, text: str) -> str:
        if not text:
            return ""
        if self.nlp is not None:
            return self._preprocess_with_stanza(text)
        return self._preprocess_simple(text)

    def _preprocess_simple(self, text: str) -> str:
        tokens = re.findall(r"[a-záčďéěíňóřšťúůýž]{3,}", text.lower())
        processed_tokens = []
        for token in tokens:
            if token in self.stop_words:
                continue
            if token.isnumeric():
                continue
            processed_tokens.append(token)
        return " ".join(processed_tokens)

    def _preprocess_with_stanza(self, text: str) -> str:
        doc = self.nlp(text)
        processed_tokens = []
        for sentence in doc.sentences:
            for word in sentence.words:
                lemma = (
                    word.lemma.lower().strip()
                    if word.lemma
                    else word.text.lower()
                )
                if len(lemma) < 3:
                    continue
                if lemma in self.stop_words:
                    continue
                if not re.match(r"^[a-záčďéěíňóřšťúůýž]+$", lemma):
                    continue
                processed_tokens.append(lemma)
        return " ".join(processed_tokens)

    def count_currency_mentions(self, text: str) -> int:
        if not text:
            return 0

        pattern = r"Kč|€|\$|CZK|EUR|USD|AUD|eur|GBP|£"

        matches = re.findall(pattern, text)
        return len(matches)
```

```python
def main_preprocessing():
    OUTPUT_DIR.mkdir(parents=True, exist_ok=True)

    print("1. Načítavam dáta...")
    try:
        df = load_jsonl(INPUT_FILE)
    except FileNotFoundError:
        print(f"Chyba: Súbor {INPUT_FILE} nebol nájdený. Skontrolujte cestu.")
        return None

    if df.empty:
        print("Dataset je prázdny.")
        return None

    required_columns = {HTML_COLUMN, LABEL_COLUMN}
    missing_columns = required_columns - set(df.columns)
    if missing_columns:
        raise ValueError(f"V datasete chýbajú stĺpce: {sorted(missing_columns)}")

    print(f"Načítaných záznamov: {len(df)}")

    processor = TextProcessor()

    print("2. Dekódujem, čistím a predspracúvam HTML...")
    df["html"] = df[HTML_COLUMN].apply(processor.decompress_html)
    df["cleaned_text"] = df["html"].apply(processor.clean_html)

    df = df[df["cleaned_text"].str.len() >= MIN_TEXT_LENGTH].reset_index(drop=True)
    print(f"Po odstránení krátkych textov: {len(df)}")

    df["currency_count"] = df["cleaned_text"].apply(processor.count_currency_mentions)

    if FILTER_NON_CZECH:
        print("3. Filtrujem iba pravdepodobne české stránky...")
        # Pre rýchlosť v notebocku môžeme filtrovať len prvých N riadkov, ak je dataset obrovský
        # Ale tu použijeme celú sadu, ak je to možné
        df["is_czech"] = df["cleaned_text"].apply(is_probably_czech)
        df = df[df["is_czech"]].reset_index(drop=True)
        print(f"Po jazykovom filtrovaní: {len(df)}")

    print("4. Spracúvam labely...")
    df["labels"] = df[LABEL_COLUMN].apply(parse_labels)
    df = df[df["labels"].map(len) > 0].reset_index(drop=True)
    df = filter_rare_labels(df, "labels", MIN_LABEL_COUNT)

    if df.empty:
        print("Po filtrovaní nezostali žiadne použiteľné dáta.")
        return None

    print("5. NLP predspracovanie textu...")
    df["processed_text"] = df["cleaned_text"].apply(processor.preprocess_text)
    df = df[df["processed_text"].str.len() >= MIN_TEXT_LENGTH].reset_index(drop=True)

    print(f"Finálny počet záznamov: {len(df)}")
    return df, processor

# Spustiť predspracovanie
result = main_preprocessing()
if result:
    df, processor = result
else:
    df = None
```

```python
if USE_LDA:
    print("9. Vytváram LDA feature-y...")

    # Initialize LDA with existing config parameters
    lda = LatentDirichletAllocation(
        n_components=NUM_TOPICS,
        max_iter=LDA_MAX_ITER,
        learning_method='batch',
        random_state=RANDOM_STATE,
        n_jobs=-1
    )

    # Fit LDA on the word TF-IDF matrix of the training set
    # Note: You might want to use X_train_word here or a combined matrix depending on your strategy.
    # Typically, LDA is fitted on the word counts/TF-IDF of the training data.
    lda.fit(X_train_word)

    # Transform train, valid, and test sets to get topic distributions
    X_train_lda = lda.transform(X_train_word)
    X_valid_lda = lda.transform(X_valid_word)
    X_test_lda = lda.transform(X_test_word)

    print(f"LDA features: {X_train_lda.shape[1]}")
else:
    # Create dummy arrays if LDA is disabled to keep stacking logic consistent
    X_train_lda = csr_matrix((X_train_word.shape[0], NUM_TOPICS))
    X_valid_lda = csr_matrix((X_valid_word.shape[0], NUM_TOPICS))
    X_test_lda = csr_matrix((X_test_word.shape[0], NUM_TOPICS))
```

```python
X_train_char = char_vectorizer.fit_transform(X_train)
X_valid_char = char_vectorizer.transform(X_valid)
X_test_char = char_vectorizer.transform(X_test)

X_train_currency = csr_matrix(df.loc[train_idx_rel, "currency_count"].values).T
X_valid_currency = csr_matrix(df.loc[valid_idx_rel, "currency_count"].values).T
X_test_currency = csr_matrix(df.loc[test_idx, "currency_count"].values).T

X_train_final = hstack([X_train_word, X_train_char, X_train_lda, X_train_currency]).tocsr()
X_valid_final = hstack([X_valid_word, X_valid_char, X_valid_lda, X_valid_currency]).tocsr()
X_test_final = hstack([X_test_word, X_test_char, X_test_lda, X_test_currency]).tocsr()

print(f"Word features: {X_train_word.shape[1]}")
print(f"Char features: {X_train_char.shape[1]}")
if USE_LDA:
    print(f"LDA features: {X_train_lda.shape[1]}")
print(f"Currency features: {X_train_currency.shape[1]}")
print(f"Celkový počet features: {X_train_final.shape[1]}")
```

```python
print("12. Evaluujem na testovacej množine...")
y_test_proba = get_positive_class_proba(classifier, X_test_final)
y_test_pred = apply_thresholds(
    y_test_proba,
    thresholds,
    ensure_at_least_one_label=ENSURE_AT_LEAST_ONE_LABEL,
)

print(
    classification_report(
        y_test,
        y_test_pred,
        target_names=mlb.classes_,
        zero_division=0,
    )
)

micro_f1 = f1_score(y_test, y_test_pred, average="micro", zero_division=0)
macro_f1 = f1_score(y_test, y_test_pred, average="macro", zero_division=0)
weighted_f1 = f1_score(
    y_test,
    y_test_pred,
    average="weighted",
    zero_division=0,
)

print(f"Micro F1:    {micro_f1:.4f}")
print(f"Macro F1:    {macro_f1:.4f}")
print(f"Weighted F1: {weighted_f1:.4f}")
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
from sklearn.preprocessing import MultiLabelBinarizer

# We'll count, for each true label, how often it was missed
# and which other labels were predicted at the same time.
labels = mlb.classes_   # e.g. ['adult', 'computers', ...]
n_labels = len(labels)

# Initialize matrix: row = true label, col = predicted label
confusion_mat = np.zeros((n_labels, n_labels), dtype=int)

# Loop over samples
for true_vec, pred_vec in zip(y_test, y_test_pred):
    true_pos = np.where(true_vec == 1)[0]
    pred_pos = np.where(pred_vec == 1)[0]

    for t_idx in true_pos:
        # If the true label was NOT predicted, record what was predicted instead
        if t_idx not in pred_pos:
            for p_idx in pred_pos:
                # If the model predicted p_idx while missing t_idx
                confusion_mat[t_idx, p_idx] += 1
        # Optionally, you could also include cases where t_idx was predicted
        # but another label was also added, but that's not a misclassification of t_idx.

# Plot heatmap
plt.figure(figsize=(10, 8))
sns.heatmap(
    confusion_mat,
    annot=True,
    fmt='d',
    cmap='Reds',
    xticklabels=labels,
    yticklabels=labels,
    square=True
)
plt.title('Multilabel confusion: true label missed → predicted label')
plt.xlabel('Predicted Label')
plt.ylabel('True Label (missed)')
plt.tight_layout()
plt.show()
```
