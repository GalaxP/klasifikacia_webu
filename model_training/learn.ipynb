```python
from pycaret.classification import ClassificationExperiment
import pandas as pd

exp = ClassificationExperiment()

backedup_df = pd.read_parquet("dataset.parquet")

clf_setup = exp.setup(
    data=backedup_df,
    target="category_encoded",
    preprocess=False,
    session_id=123,
    use_gpu=False,
    fold=3,
    fold_strategy="kfold",
    fold_shuffle=True,
    verbose=True
)

best = exp.compare_models(
    exclude=["gbc"]
)
```
