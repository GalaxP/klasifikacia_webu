```python
import base64
import gzip
import json
import re
import warnings
from collections import Counter
from pathlib import Path
from typing import Any, Dict, List, Optional, Sequence, Tuple

import joblib
import numpy as np
import pandas as pd
from bs4 import BeautifulSoup, Comment
from lightgbm import LGBMClassifier
from scipy.sparse import csr_matrix, hstack
from sklearn.decomposition import LatentDirichletAllocation
from sklearn.feature_extraction.text import CountVectorizer, TfidfVectorizer
from sklearn.metrics import classification_report, f1_score
from sklearn.model_selection import train_test_split
from sklearn.multiclass import OneVsRestClassifier
from sklearn.preprocessing import MultiLabelBinarizer

warnings.filterwarnings("ignore")

# -----------------------------------------------------------------------------
# Konfigurácia
# -----------------------------------------------------------------------------

INPUT_FILE = r"D:\dataset_cz_v1_final.jsonl"

HTML_COLUMN = "compressed_html"
LABEL_COLUMN = "categories"

OUTPUT_DIR = Path("model_output_cz_multilabel")

RANDOM_STATE = 42
TEST_SIZE = 0.2
VALID_SIZE = 0.15

MIN_TEXT_LENGTH = 120
MIN_LABEL_COUNT = 5

MAX_FEATURES = 80_000
TFIDF_MIN_DF = 2
TFIDF_MAX_DF = 0.9
NGRAM_RANGE = (1, 2)

USE_LDA = True
NUM_TOPICS = 30
LDA_MAX_FEATURES = 20_000
LDA_MAX_ITER = 10
TOP_N_WORDS = 20

FILTER_NON_CZECH = True
MIN_CZECH_PROB = 0.75

USE_SPACY_LEMMATIZATION = False
USE_STANZA_LEMMATIZATION = True
USE_LEMMATIZATION = True
SPACY_MODEL = "cs_core_news_sm"

ENSURE_AT_LEAST_ONE_LABEL = True

THRESHOLD_GRID = np.arange(0.1, 0.91, 0.05)


CZECH_STOPWORDS = {
    "a",
    "aby",
    "ahoj",
    "aj",
    "ale",
    "anebo",
    "ani",
    "ano",
    "asi",
    "aspoň",
    "atd",
    "atp",
    "až",
    "bez",
    "bude",
    "budem",
    "budeme",
    "budeš",
    "budete",
    "budou",
    "budu",
    "by",
    "byl",
    "byla",
    "byli",
    "bylo",
    "byly",
    "bys",
    "čau",
    "či",
    "čím",
    "co",
    "což",
    "cz",
    "další",
    "dnes",
    "do",
    "dobrý",
    "ho",
    "i",
    "jak",
    "jakmile",
    "jako",
    "je",
    "jeho",
    "jej",
    "její",
    "jejich",
    "jemu",
    "jen",
    "jenom",
    "jenž",
    "ještě",
    "ji",
    "jiné",
    "jsem",
    "jsi",
    "jsme",
    "jsou",
    "jste",
    "já",
    "jí",
    "jím",
    "jíž",
    "k",
    "kam",
    "kde",
    "kdo",
    "když",
    "ke",
    "která",
    "které",
    "kterou",
    "který",
    "kteří",
    "ku",
    "má",
    "mám",
    "máme",
    "máte",
    "mé",
    "mě",
    "mi",
    "mne",
    "mnou",
    "moc",
    "mohl",
    "mohou",
    "moje",
    "moji",
    "můj",
    "musí",
    "my",
    "na",
    "nad",
    "nám",
    "námi",
    "např",
    "nás",
    "náš",
    "naše",
    "naši",
    "ne",
    "nebo",
    "nebyl",
    "nebyla",
    "nebyli",
    "nebylo",
    "nechť",
    "nedělá",
    "něco",
    "nějak",
    "někde",
    "někdo",
    "nemají",
    "nemáme",
    "nemáte",
    "není",
    "než",
    "nic",
    "nich",
    "ním",
    "nímž",
    "nyní",
    "o",
    "od",
    "ode",
    "on",
    "ona",
    "oni",
    "ono",
    "ony",
    "pak",
    "po",
    "pod",
    "podle",
    "pokud",
    "pouze",
    "právě",
    "pro",
    "proč",
    "proto",
    "protože",
    "před",
    "přes",
    "při",
    "s",
    "se",
    "si",
    "sk",
    "snad",
    "spíše",
    "sta",
    "sté",
    "sto",
    "ta",
    "tady",
    "tak",
    "takhle",
    "taky",
    "tam",
    "tamhle",
    "tato",
    "tebe",
    "tebou",
    "tedy",
    "ten",
    "tento",
    "teto",
    "ti",
    "tisíc",
    "to",
    "tobě",
    "tohle",
    "toho",
    "tohoto",
    "tom",
    "tomto",
    "tomu",
    "tomuto",
    "toto",
    "trošku",
    "tu",
    "tuto",
    "tvá",
    "tvé",
    "tvoje",
    "tvůj",
    "ty",
    "tyto",
    "u",
    "určitě",
    "už",
    "v",
    "vám",
    "vámi",
    "vás",
    "váš",
    "vaše",
    "vaši",
    "ve",
    "vedle",
    "více",
    "však",
    "všechen",
    "všechno",
    "všichni",
    "vy",
    "z",
    "za",
    "zatímco",
    "zda",
    "zde",
    "ze",
    "že",
    "pro",
    "cookies",
    "soubor",
    "souhlas",
    "lišta",
    "nastavení",
}


# -----------------------------------------------------------------------------
# Pomocné funkcie
# -----------------------------------------------------------------------------

def load_jsonl(filepath: str) -> pd.DataFrame:
    records: List[Dict[str, Any]] = []

    with open(filepath, "r", encoding="utf-8") as file:
        for line_number, line in enumerate(file, start=1):
            line = line.strip()

            if not line:
                continue

            try:
                records.append(json.loads(line))
            except json.JSONDecodeError:
                print(f"Preskakujem chybný JSON na riadku {line_number}.")

    return pd.DataFrame(records)


def parse_labels(value: Any) -> List[str]:
    if value is None:
        return []

    if isinstance(value, list):
        labels = value
    elif isinstance(value, str):
        value = value.strip()

        if not value:
            return []

        if value.startswith("["):
            try:
                parsed = json.loads(value)
                labels = parsed if isinstance(parsed, list) else [value]
            except json.JSONDecodeError:
                labels = re.split(r"[,;|]", value)
        else:
            labels = re.split(r"[,;|]", value)
    else:
        labels = [str(value)]

    cleaned_labels = []

    for label in labels:
        label = str(label).strip().lower()

        if label:
            cleaned_labels.append(label)

    return sorted(set(cleaned_labels))


def filter_rare_labels(
    df: pd.DataFrame,
    label_col: str,
    min_count: int,
) -> pd.DataFrame:
    label_counts = Counter(
        label for labels in df[label_col] for label in labels
    )

    valid_labels = {
        label for label, count in label_counts.items() if count >= min_count
    }

    df = df.copy()
    df[label_col] = df[label_col].apply(
        lambda labels: sorted(label for label in labels if label in valid_labels)
    )
    df = df[df[label_col].map(len) > 0].reset_index(drop=True)

    print(f"Počet labelov po filtrovaní: {len(valid_labels)}")
    print(f"Odstránené labely s výskytom menším ako {min_count}.")

    return df


def multilabel_train_test_split(
    X: Sequence[Any],
    y: np.ndarray,
    test_size: float,
    random_state: int,
) -> Tuple[np.ndarray, np.ndarray]:
    indices = np.arange(len(X))

    try:
        from iterstrat.ml_stratifiers import MultilabelStratifiedShuffleSplit

        splitter = MultilabelStratifiedShuffleSplit(
            n_splits=1,
            test_size=test_size,
            random_state=random_state,
        )

        train_idx, test_idx = next(splitter.split(np.zeros(len(indices)), y))
        return train_idx, test_idx

    except ImportError:
        print(
            "Balík iterative-stratification nie je nainštalovaný. "
            "Používam obyčajný train_test_split."
        )

        train_idx, test_idx = train_test_split(
            indices,
            test_size=test_size,
            random_state=random_state,
        )

        return train_idx, test_idx


def get_positive_class_proba(classifier: OneVsRestClassifier, X: Any) -> np.ndarray:
    probabilities = classifier.predict_proba(X)

    if isinstance(probabilities, list):
        probabilities = np.vstack([item[:, 1] for item in probabilities]).T

    return np.asarray(probabilities)


def tune_thresholds(
    y_true: np.ndarray,
    y_proba: np.ndarray,
    threshold_grid: np.ndarray,
) -> np.ndarray:
    thresholds = np.full(y_true.shape[1], 0.5, dtype=float)

    for label_idx in range(y_true.shape[1]):
        if y_true[:, label_idx].sum() == 0:
            continue

        best_threshold = 0.5
        best_score = -1.0

        for threshold in threshold_grid:
            y_pred_label = (y_proba[:, label_idx] >= threshold).astype(int)
            score = f1_score(
                y_true[:, label_idx],
                y_pred_label,
                zero_division=0,
            )

            if score > best_score:
                best_score = score
                best_threshold = threshold

        thresholds[label_idx] = best_threshold

    return thresholds


def apply_thresholds(
    y_proba: np.ndarray,
    thresholds: np.ndarray,
    ensure_at_least_one_label: bool = True,
) -> np.ndarray:
    y_pred = (y_proba >= thresholds).astype(int)

    if ensure_at_least_one_label:
        empty_rows = np.where(y_pred.sum(axis=1) == 0)[0]

        if len(empty_rows) > 0:
            best_label_indices = np.argmax(y_proba[empty_rows], axis=1)
            y_pred[empty_rows, best_label_indices] = 1

    return y_pred


def is_probably_czech(text: str) -> bool:
    if not FILTER_NON_CZECH:
        return True

    try:
        from langdetect import DetectorFactory, detect_langs

        DetectorFactory.seed = RANDOM_STATE

        sample = text[:5000]
        detected_languages = detect_langs(sample)

        return any(
            item.lang == "cs" and item.prob >= MIN_CZECH_PROB
            for item in detected_languages
        )

    except ImportError:
        print("Balík langdetect nie je nainštalovaný. Jazykový filter sa vypína.")
        return True
    except Exception:
        return False


# -----------------------------------------------------------------------------
# Spracovanie HTML a textu
# -----------------------------------------------------------------------------

class TextProcessor:
    def __init__(self) -> None:
        self.stop_words = CZECH_STOPWORDS
        self.nlp = None

        if USE_STANZA_LEMMATIZATION:
            try:
                import stanza

                stanza.download("cs", verbose=False)
                self.nlp = stanza.Pipeline(
                    lang="cs",
                    processors="tokenize,pos,lemma",
                    use_gpu=False,
                    verbose=False,
                )
                print("Používam Stanza český model na lematizáciu.")
            except Exception as exc:
                print(
                    "Stanza český model sa nepodarilo načítať. "
                    "Pokračujem bez lematizácie."
                )
                print(f"Detail: {exc}")
                self.nlp = None

    def decompress_html(self, value: Any) -> str:
        if value is None:
            return ""

        if not isinstance(value, str):
            value = str(value)

        value = value.strip()

        if not value:
            return ""

        if "<html" in value.lower() or "<body" in value.lower() or "<div" in value.lower():
            return value

        try:
            decoded = base64.b64decode(value)
        except Exception:
            return value

        try:
            decompressed = gzip.decompress(decoded)
            return decompressed.decode("utf-8", errors="ignore")
        except Exception:
            try:
                return decoded.decode("utf-8", errors="ignore")
            except Exception:
                return ""

    def clean_html(self, html_text: str) -> str:
        if not html_text:
            return ""

        soup = BeautifulSoup(html_text, "lxml")

        for comment in soup.find_all(string=lambda item: isinstance(item, Comment)):
            comment.extract()

        for tag in soup(
            [
                "script",
                "style",
                "noscript",
                "svg",
                "canvas",
                "iframe",
                "form",
                "nav",
                "footer",
                "aside",
            ]
        ):
            tag.decompose()

        important_parts: List[str] = []

        if soup.title and soup.title.string:
            important_parts.extend([soup.title.string] * 3)

        for meta_name in ["description", "keywords"]:
            meta = soup.find("meta", attrs={"name": meta_name})

            if meta and meta.get("content"):
                important_parts.extend([meta["content"]] * 2)

        for heading in soup.find_all(["h1", "h2", "h3"]):
            heading_text = heading.get_text(" ", strip=True)

            if heading_text:
                important_parts.extend([heading_text] * 2)

        body_text = soup.get_text(separator=" ", strip=True)

        text = " ".join(important_parts + [body_text])
        text = text.lower()

        text = re.sub(r"https?://\S+|www\.\S+", " ", text)
        text = re.sub(r"\S+@\S+", " ", text)
        text = re.sub(r"&nbsp;", " ", text)

        # Zachová české znaky, čísla a medzery.
        text = re.sub(r"[^0-9a-záčďéěíňóřšťúůýž\s-]", " ", text)
        text = re.sub(r"\s+", " ", text).strip()

        return text

    def preprocess_text(self, text: str) -> str:
        if not text:
            return ""

        if self.nlp is not None:
            return self._preprocess_with_stanza(text)

        return self._preprocess_simple(text)

    def _preprocess_simple(self, text: str) -> str:
        tokens = re.findall(r"[a-záčďéěíňóřšťúůýž]{3,}", text.lower())
        processed_tokens = []

        for token in tokens:
            if token in self.stop_words:
                continue

            if token.isnumeric():
                continue

            processed_tokens.append(token)

        return " ".join(processed_tokens)

    def _preprocess_with_stanza(self, text: str) -> str:
        doc = self.nlp(text)
        processed_tokens = []

        for sentence in doc.sentences:
            for word in sentence.words:
                lemma = word.lemma.lower().strip() if word.lemma else word.text.lower()

                if len(lemma) < 3:
                    continue

                if lemma in self.stop_words:
                    continue

                if not re.match(r"^[a-záčďéěíňóřšťúůýž]+$", lemma):
                    continue

                processed_tokens.append(lemma)

        return " ".join(processed_tokens)

# -----------------------------------------------------------------------------
# Hlavný tréning
# -----------------------------------------------------------------------------

def main() -> None:
    OUTPUT_DIR.mkdir(parents=True, exist_ok=True)

    print("1. Načítavam dáta...")
    df = load_jsonl(INPUT_FILE)

    if df.empty:
        print("Dataset je prázdny.")
        return

    required_columns = {HTML_COLUMN, LABEL_COLUMN}
    missing_columns = required_columns - set(df.columns)

    if missing_columns:
        raise ValueError(f"V datasete chýbajú stĺpce: {sorted(missing_columns)}")

    print(f"Načítaných záznamov: {len(df)}")

    processor = TextProcessor()

    print("2. Dekódujem, čistím a predspracúvam HTML...")
    df["html"] = df[HTML_COLUMN].apply(processor.decompress_html)
    df["cleaned_text"] = df["html"].apply(processor.clean_html)

    df = df[df["cleaned_text"].str.len() >= MIN_TEXT_LENGTH].reset_index(drop=True)
    print(f"Po odstránení krátkych textov: {len(df)}")

    if FILTER_NON_CZECH:
        print("3. Filtrujem iba pravdepodobne české stránky...")
        df["is_czech"] = df["cleaned_text"].apply(is_probably_czech)
        df = df[df["is_czech"]].reset_index(drop=True)
        print(f"Po jazykovom filtrovaní: {len(df)}")

    print("4. Spracúvam labely...")
    df["labels"] = df[LABEL_COLUMN].apply(parse_labels)
    df = df[df["labels"].map(len) > 0].reset_index(drop=True)
    df = filter_rare_labels(df, "labels", MIN_LABEL_COUNT)

    if df.empty:
        print("Po filtrovaní nezostali žiadne použiteľné dáta.")
        return

    print("5. NLP predspracovanie textu...")
    df["processed_text"] = df["cleaned_text"].apply(processor.preprocess_text)
    df = df[df["processed_text"].str.len() >= MIN_TEXT_LENGTH].reset_index(drop=True)

    print(f"Finálny počet záznamov: {len(df)}")

    mlb = MultiLabelBinarizer()
    y = mlb.fit_transform(df["labels"])

    X_text = df["processed_text"].to_numpy()

    print(f"Počet multilabel kategórií: {len(mlb.classes_)}")
    print(f"Kategórie: {list(mlb.classes_)}")

    print("6. Delím dáta na train, valid a test...")
    train_valid_idx, test_idx = multilabel_train_test_split(
        X_text,
        y,
        test_size=TEST_SIZE,
        random_state=RANDOM_STATE,
    )

    X_train_valid = X_text[train_valid_idx]
    y_train_valid = y[train_valid_idx]

    X_test = X_text[test_idx]
    y_test = y[test_idx]

    train_idx_rel, valid_idx_rel = multilabel_train_test_split(
        X_train_valid,
        y_train_valid,
        test_size=VALID_SIZE,
        random_state=RANDOM_STATE,
    )

    X_train = X_train_valid[train_idx_rel]
    y_train = y_train_valid[train_idx_rel]

    X_valid = X_train_valid[valid_idx_rel]
    y_valid = y_train_valid[valid_idx_rel]

    print(f"Train: {len(X_train)}")
    print(f"Valid: {len(X_valid)}")
    print(f"Test: {len(X_test)}")

    print("7. Vytváram TF-IDF feature-y...")
    tfidf_vectorizer = TfidfVectorizer(
        max_features=MAX_FEATURES,
        min_df=TFIDF_MIN_DF,
        max_df=TFIDF_MAX_DF,
        ngram_range=NGRAM_RANGE,
        sublinear_tf=True,
        norm="l2",
        dtype=np.float32,
    )

    X_train_tfidf = tfidf_vectorizer.fit_transform(X_train)
    X_valid_tfidf = tfidf_vectorizer.transform(X_valid)
    X_test_tfidf = tfidf_vectorizer.transform(X_test)

    feature_blocks_train = [X_train_tfidf]
    feature_blocks_valid = [X_valid_tfidf]
    feature_blocks_test = [X_test_tfidf]

    count_vectorizer: Optional[CountVectorizer] = None
    lda_model: Optional[LatentDirichletAllocation] = None

    if USE_LDA:
        print("8. Vytváram LDA topic feature-y...")
        count_vectorizer = CountVectorizer(
            max_features=LDA_MAX_FEATURES,
            min_df=TFIDF_MIN_DF,
            max_df=TFIDF_MAX_DF,
        )

        X_train_counts = count_vectorizer.fit_transform(X_train)
        X_valid_counts = count_vectorizer.transform(X_valid)
        X_test_counts = count_vectorizer.transform(X_test)

        lda_model = LatentDirichletAllocation(
            n_components=NUM_TOPICS,
            max_iter=LDA_MAX_ITER,
            learning_method="online",
            random_state=RANDOM_STATE,
            n_jobs=-1,
        )

        X_train_lda = lda_model.fit_transform(X_train_counts)
        X_valid_lda = lda_model.transform(X_valid_counts)
        X_test_lda = lda_model.transform(X_test_counts)

        feature_blocks_train.append(csr_matrix(X_train_lda))
        feature_blocks_valid.append(csr_matrix(X_valid_lda))
        feature_blocks_test.append(csr_matrix(X_test_lda))

        feature_names = count_vectorizer.get_feature_names_out()

        print("\nTop slová pre LDA témy:")
        for topic_idx, topic in enumerate(lda_model.components_):
            top_indices = topic.argsort()[:-TOP_N_WORDS - 1:-1]
            top_words = [feature_names[index] for index in top_indices]
            print(f"Téma {topic_idx + 1}: {', '.join(top_words)}")

    X_train_final = hstack(feature_blocks_train).tocsr()
    X_valid_final = hstack(feature_blocks_valid).tocsr()
    X_test_final = hstack(feature_blocks_test).tocsr()

    print(f"\nCelkový počet feature-ov: {X_train_final.shape[1]}")

    print("9. Trénujem LightGBM One-vs-Rest multilabel model...")

    base_model = LGBMClassifier(
        objective="binary",
        n_estimators=500,
        learning_rate=0.03,
        num_leaves=63,
        max_depth=-1,
        subsample=0.85,
        subsample_freq=1,
        colsample_bytree=0.85,
        reg_alpha=0.1,
        reg_lambda=1.0,
        min_child_samples=20,
        class_weight="balanced",
        random_state=RANDOM_STATE,
        n_jobs=-1,
        verbose=-1,
    )

    classifier = OneVsRestClassifier(
        base_model,
        n_jobs=-1,
    )

    classifier.fit(X_train_final, y_train)

    print("10. Ladím thresholdy na validačnej množine...")
    y_valid_proba = get_positive_class_proba(classifier, X_valid_final)

    thresholds = tune_thresholds(
        y_valid,
        y_valid_proba,
        THRESHOLD_GRID,
    )

    print("Thresholdy podľa labelov:")
    for label, threshold in zip(mlb.classes_, thresholds):
        print(f"  {label}: {threshold:.2f}")

    print("11. Evaluujem na testovacej množine...")
    y_test_proba = get_positive_class_proba(classifier, X_test_final)
    y_test_pred = apply_thresholds(
        y_test_proba,
        thresholds,
        ensure_at_least_one_label=ENSURE_AT_LEAST_ONE_LABEL,
    )

    print(
        classification_report(
            y_test,
            y_test_pred,
            target_names=mlb.classes_,
            zero_division=0,
        )
    )

    micro_f1 = f1_score(y_test, y_test_pred, average="micro", zero_division=0)
    macro_f1 = f1_score(y_test, y_test_pred, average="macro", zero_division=0)
    weighted_f1 = f1_score(y_test, y_test_pred, average="weighted", zero_division=0)

    print(f"Micro F1:    {micro_f1:.4f}")
    print(f"Macro F1:    {macro_f1:.4f}")
    print(f"Weighted F1: {weighted_f1:.4f}")

    print("12. Ukladám model a artefakty...")

    bundle = {
        "classifier": classifier,
        "tfidf_vectorizer": tfidf_vectorizer,
        "count_vectorizer": count_vectorizer,
        "lda_model": lda_model,
        "mlb": mlb,
        "thresholds": thresholds,
        "config": {
            "html_column": HTML_COLUMN,
            "label_column": LABEL_COLUMN,
            "use_lda": USE_LDA,
            "use_lemmatization": USE_LEMMATIZATION,
            "use_stanza_lemmatization": USE_STANZA_LEMMATIZATION,
            "ensure_at_least_one_label": ENSURE_AT_LEAST_ONE_LABEL,
        },
    }

    joblib.dump(bundle, OUTPUT_DIR / "cz_multilabel_html_classifier.joblib")

    with open(OUTPUT_DIR / "labels.json", "w", encoding="utf-8") as file:
        json.dump(list(mlb.classes_), file, ensure_ascii=False, indent=2)

    print(f"Hotovo. Model uložený do: {OUTPUT_DIR}")


if __name__ == "__main__":
    main()
```
